# RSA

## Matemático

```python
# Geração de chaves
p = 61
q = 53
n = p * q                      # Módulo da chave pública
phi = (p - 1) * (q - 1)        

e = 17                         # Expoente público

# Calculamos 'd' (expoente privado)
d = pow(e, -1, phi)            # chave privada

print(f"Chave Pública: (n={n}, e={e})")
print(f"Chave Privada: (d={d})\n")

# Criptografar
mensagem_original = 121          # A mensagem deve ser um número menor que 'n'
print(f"Mensagem original: {mensagem_original}")

# Usamos a chave pública: c = m^e mod n
texto_criptografado = pow(mensagem_original, e, n)
print(f"Texto criptografado: {texto_criptografado}")

# Usamos a chave privada: m = c^d mod n
mensagem_recuperado = pow(texto_criptografado, d, n)
print(f"Mensaje descifrado: {mensagem_recuperado}")
```

## Biblioteca cryptography

```python
from cryptography.hazmat.primitives.asymmetric import rsa
from cryptography.hazmat.primitives.asymmetric import padding
from cryptography.hazmat.primitives import hashes

# Geração de chaves Seguras (2048 bits)
chave_privada = rsa.generate_private_key(
    public_exponent=65537,     # 65537 es el estándar matemático para 'e'
    key_size=2048              # Tamaño de 'n'
)
chave_publica = chave_privada.public_key()

# Mensagem a criptografar (deve ser convertido a bytes)
mensagem = b'Um segredo corporativo.'

# Usando a chave pública
# Vamos usar OAEP (Optimal Asymmetric Encryption Padding) para máxima seguridade
texto_criptografado = chave_publica.encrypt(
    mensagem,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)

print(f"Mensagem criptografado (em bytes ilegíveis): {texto_criptografado[:50]}...")

# Usamos a chave privada)
texto_decodificado = chave_privada.decrypt(
    texto_criptografado,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)

print(f"Mensajem recuperada: {texto_decodificado.decode('utf-8')}")
```
